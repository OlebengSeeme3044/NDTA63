## Introduction

This section presents visualisations of South Africa's ICT regulatory and cybersecurity performance using the cleaned, analysis-ready datasets. The purpose of the visualisations is to identify trends, compare dimensions, examine changes over time, and assess whether the available data supports a meaningful relationship analysis between ICT regulation and cybersecurity readiness.

The ICT analysis covers the Overall ICT Regulatory Score and its regulatory dimensions across the available period from 2016 to 2022. The GCI analysis compares the available cybersecurity assessment years, 2020 and 2024, and examines changes across its dimensions.

The visualisations are supported by the numerical findings produced during the quantitative analysis. However, differences in the temporal coverage of the ICT and GCI datasets are considered when interpreting their relationship. Where the data structure does not support a statistically meaningful comparison, the limitation is explicitly identified rather than forcing an unsupported statistical relationship.

### ICT Regulatory Tracker

```python
import pandas as pd
import matplotlib.pyplot as plt

ict = pd.read_excel("ict_regulatory_cleaned (2).xlsx")

ict.head()
```

```python
ict[[
    "year",
    "overall_ict_score",
    "regulatory_authority",
    "regulatory_mandate",
    "regulatory_regime",
    "competition_framework"
]]
```

```python

plt.figure(figsize=(10, 6))

plt.plot(
    ict["year"],
    ict["overall_ict_score"],
    marker="o",
    linewidth=2.5,
    markersize=7
)

for x, y in zip(ict["year"], ict["overall_ict_score"]):
    plt.annotate(
        f"{y:.1f}",
        (x, y),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center",
        fontsize=10
    )

plt.title(
    "South Africa ICT Regulatory Tracker Overall Score, 2016–2022",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Year", fontsize=11)
plt.ylabel("Overall ICT Regulatory Score", fontsize=11)

plt.xticks(ict["year"])
plt.ylim(66, 86)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

### Analysis

The ICT Regulatory Tracker Overall Score shows a consistent upward trend in South Africa from 2016 to 2022. The score increased from 68.5 in 2016 to 84.0 in 2022, with no year-to-year decline. This indicates an overall strengthening of the country's measured ICT regulatory environment during the period.

Overall, the score increased by 15.5 points between 2016 and 2022, representing an increase of approximately 22.63%. The largest year-on-year increase occurred between 2018 and 2019, when the score rose from 73.5 to 77.0, an increase of 3.5 points.


```python
plt.figure(figsize=(11, 6))

plt.plot(
    ict["year"],
    ict["regulatory_authority"],
    marker="o",
    linewidth=2,
    label="Regulatory Authority"
)

plt.plot(
    ict["year"],
    ict["regulatory_mandate"],
    marker="o",
    linewidth=2,
    label="Regulatory Mandate"
)

plt.plot(
    ict["year"],
    ict["regulatory_regime"],
    marker="o",
    linewidth=2,
    label="Regulatory Regime"
)

plt.plot(
    ict["year"],
    ict["competition_framework"],
    marker="o",
    linewidth=2,
    label="Competition Framework"
)

plt.title(
    "South Africa ICT Regulatory Dimensions, 2016–2022",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Year", fontsize=11)
plt.ylabel("Regulatory Score", fontsize=11)

plt.xticks(ict["year"])

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.legend()

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

### Analysis:
The four ICT regulatory dimensions all show a steady upward trend in South Africa between 2016 and 2022. Regulatory Authority increased from 14 to 18, while Regulatory Mandate increased from 16.5 to 19.5. Regulatory Regime recorded the largest improvement, rising from 20 to 25, followed by Competition Framework, which increased from 18 to 21.5. Overall, the chart shows that improvements were broad-based across all four regulatory dimensions, with Regulatory Regime showing the greatest absolute increase.

```python
import numpy as np
import matplotlib.pyplot as plt

gci_scores = gci[
    gci["INDICATOR_LABEL"].str.contains(
        "Score",
        case=False,
        na=False
    )
].copy()

gci_scores = gci_scores[
    gci_scores["INDICATOR"] != "ITU_GCI_GCI_OVRL_SCRE"
]

def get_dimension(label):
    label = str(label).lower()

    if "legal" in label:
        return "Legal"
    elif "technical" in label:
        return "Technical"
    elif "organizational" in label:
        return "Organizational"
    elif "capacity development" in label:
        return "Capacity Development"
    elif "cooperative" in label:
        return "Cooperative"
    else:
        return None

gci_scores["Dimension"] = gci_scores["INDICATOR_LABEL"].apply(get_dimension)

gci_scores = gci_scores.dropna(subset=["Dimension"])

dimension_data = gci_scores.pivot(
    index="Dimension",
    columns="YEAR",
    values="VALUE"
)

dimension_order = [
    "Legal",
    "Technical",
    "Organizational",
    "Capacity Development",
    "Cooperative"
]

dimension_data = dimension_data.reindex(dimension_order)

print(dimension_data)
```

```python
x = np.arange(len(dimension_data.index))
width = 0.35

plt.figure(figsize=(11, 6))

bars_2020 = plt.bar(
    x - width / 2,
    dimension_data[2020],
    width,
    label="2020"
)

bars_2024 = plt.bar(
    x + width / 2,
    dimension_data[2024],
    width,
    label="2024"
)

plt.title(
    "South Africa GCI Dimension Scores: 2020 vs 2024",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("GCI Dimension")
plt.ylabel("Score")

plt.xticks(
    x,
    dimension_data.index,
    rotation=15
)

plt.legend()

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

for bars in [bars_2020, bars_2024]:
    for bar in bars:
        height = bar.get_height()

        if not np.isnan(height):
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height + 0.2,
                f"{height:.1f}",
                ha="center",
                fontsize=9
            )

plt.tight_layout()
plt.show()
```

### Analysis

The comparison of South Africa's GCI dimension scores between 2020 and 2024 shows that changes were not uniform across all cybersecurity dimensions. Legal, Technical, Organizational and Capacity Development scores increased, while the Cooperative dimension declined.

Technical performance increased from 15.9 in 2020 to 18.4 in 2024, while Capacity Development increased from 15.4 to 17.6. Legal increased from 17.2 to 18.9, and Organizational increased from 15.2 to 15.8. In contrast, the Cooperative dimension decreased from 18.8 to 16.2.

Overall, the chart shows that the increase in South Africa's GCI performance was driven by improvements in several dimensions, despite a decline in the Cooperative dimension.

```python
gci_change = (
    dimension_data[2024] - dimension_data[2020]
).dropna()

gci_change = gci_change.sort_values()

plt.figure(figsize=(10, 6))

bars = plt.barh(
    gci_change.index,
    gci_change.values
)

plt.axvline(
    0,
    linewidth=1
)

plt.title(
    "Change in South Africa GCI Dimension Scores, 2020–2024",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Change in Score (2024 − 2020)")
plt.ylabel("GCI Dimension")

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

for bar, value in zip(bars, gci_change.values):

    if value >= 0:
        plt.text(
            value + 0.1,
            bar.get_y() + bar.get_height() / 2,
            f"+{value:.1f}",
            va="center"
        )
    else:
        plt.text(
            value - 0.05,
            bar.get_y() + bar.get_height() / 2,
            f"{value:.1f}",
            va="center",
            ha="right"
        )

plt.tight_layout()
plt.show()
```

### Analysis

The change analysis shows how each GCI dimension shifted between 2020 and 2024. Positive values indicate improvement, while negative values indicate a decline. The chart makes it easier to identify which cybersecurity dimensions experienced the greatest change over the period. Overall, the results show that changes in cybersecurity performance were not uniform across all dimensions, highlighting areas of both progress and potential weakness.

Technical capability recorded the largest improvement at 2.48 points, followed by Capacity Development at 2.20 points and Legal at 1.72 points. In contrast, Cooperation declined by 2.64 points, making it the only dimension to show a negative change.

### Relationship Analysis:

A relationship analysis between the ICT Regulatory Tracker and the Global Cybersecurity Index was considered. However, the ICT dataset contains annual observations from 2016 to 2022, while the available GCI data contains observations for 2020 and 2024. Therefore, the datasets have only one overlapping year, 2020.

```python
years = [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

ict_available = [1, 1, 1, 1, 1, 1, 1, 0, 0]
gci_available = [0, 0, 0, 0, 1, 0, 0, 0, 1]

x = np.arange(len(years))
width = 0.35

plt.figure(figsize=(11, 6))

bars_ict = plt.bar(
    x - width / 2,
    ict_available,
    width,
    label="ICT Regulatory Tracker"
)

bars_gci = plt.bar(
    x + width / 2,
    gci_available,
    width,
    label="Global Cybersecurity Index (GCI)"
)

plt.title(
    "Availability and Temporal Overlap of ICT and GCI Observations, 2016–2024",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlabel("Year", fontsize=11)
plt.ylabel("Observation Available", fontsize=11)

plt.xticks(x, years)
plt.yticks([0, 1], ["No", "Yes"])

plt.legend()

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

overlap_index = years.index(2020)

plt.axvline(
    overlap_index,
    linestyle="--",
    linewidth=1,
    alpha=0.5
)

plt.text(
    overlap_index,
    1.08,
    "Only overlapping year",
    ha="center",
    fontsize=10
)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.ylim(0, 1.2)

plt.tight_layout()
plt.show()
```

### Relationship Analysis: Dataset Overlap and Limitations

The chart compares the availability of observations in the ICT Regulatory Tracker and the Global Cybersecurity Index (GCI) from 2016 to 2024. The ICT dataset provides annual observations from 2016 through 2022, while the available GCI data contains observations for 2020 and 2024. As a result, the two datasets overlap in only one year: 2020.

The ICT dataset therefore provides considerably greater longitudinal coverage than the available GCI data. There are no corresponding GCI observations for the ICT observations from 2016–2019 or 2021–2022. Similarly, the 2024 GCI observation cannot be directly matched with an ICT observation because the ICT dataset ends in 2022. Neither dataset contains an observation for 2023.

The numerical analysis indicates that both measures improved over their respective periods. The ICT Regulatory Tracker increased by 15.5 points (22.63%) between 2016 and 2022, while the GCI overall score increased by 7.34 points (10.34%) between 2020 and 2024. These results indicate improvement in both areas, but the different assessment periods and limited overlap mean that the changes cannot be interpreted as evidence of a statistical relationship between ICT regulation and cybersecurity readiness.

Because only one year, 2020, is common to both datasets, there are insufficient matched observations to calculate a meaningful correlation or regression between ICT regulatory performance and GCI performance. Although both datasets show improvement across their respective periods, this cannot be interpreted as evidence of a statistical relationship or causation.

The overlap analysis therefore identifies an important data limitation rather than forcing an unsupported relationship analysis. A longer period of comparable annual observations would be required to statistically assess the relationship between ICT regulatory performance and cybersecurity readiness.

### Distribution of ICT Regulatory Scores Analysis

The box plot summarises South Africa’s seven annual ICT Overall Regulatory Scores from 2016 to 2022. It provides a compact view of the central tendency, spread, and overall range of regulatory performance during the period.

The distribution should be interpreted alongside the ICT trend shown in Figure 1. The line chart demonstrates that the overall ICT regulatory score increased over time, while the box plot summarises how the seven yearly observations are distributed across the period. Because the observations represent consecutive years rather than independent samples, the box plot is primarily descriptive and should not be interpreted as evidence of a broader population distribution.

The relatively small number of observations also limits the statistical conclusions that can be drawn from the distribution. Nevertheless, the box plot is useful because it provides another perspective on the consistency and variability of South Africa’s ICT regulatory performance. When considered together with the upward trend in Figure 1, it helps show both the overall movement and the spread of the annual scores.

Overall, the distribution visualisation complements the time-series analysis by providing a summary of the ICT regulatory scores across 2016–2022, while avoiding unsupported statistical inference from the small dataset.

## Conclusion

The visualisations provide complementary evidence on South Africa's ICT regulatory and cybersecurity performance. The ICT trend analysis shows an overall improvement in the Overall ICT Regulatory Score between 2016 and 2022, while the dimension analysis shows that progress occurred across the regulatory areas, with the Regulatory Regime recording the largest absolute increase.

The GCI analysis similarly indicates an improvement in the overall score between 2020 and 2024. However, the dimension-level results demonstrate that this improvement was not uniform. Technical, Capacity Development and Legal dimensions improved, while the Cooperative dimension declined.

The distribution analysis provides an additional descriptive view of the ICT scores and complements the time-series trend by showing the spread of the annual observations.

The relationship analysis identified an important limitation in comparing the two datasets. ICT observations are available annually from 2016 to 2022, whereas the available GCI observations are limited to 2020 and 2024. Since only 2020 overlaps, there are insufficient matched observations to conduct a statistically meaningful correlation or regression analysis. The observed improvements in ICT regulation and cybersecurity readiness should therefore be interpreted as complementary findings rather than evidence that improvement in one measure caused improvement in the other.

Overall, the visualisations demonstrate changes over time, differences between dimensions, the distribution of ICT regulatory performance, and the limitations associated with comparing datasets with different temporal coverage. This provides a clearer and more evidence-based interpretation of South Africa's ICT regulatory and cybersecurity performance.
