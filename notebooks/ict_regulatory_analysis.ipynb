# Member 2: ICT Regulatory Tracker Analysis Pipeline
**Dataset:** World Bank TCdata360 / ITU ICT Regulatory Tracker  
**Focus:** South Africa (ISO3: `ZAF`)  
**Deliverables:** `ict_regulatory_cleaned.csv`, `ict_regulatory_analysis.ipynb`, `ict_regulatory_summary.docx`

```python
import numpy as np
import pandas as pd

raw_tracker_data = {
    'country_name': ['South Africa'] * 7,
    'iso3': ['ZAF'] * 7,
    'year': [2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'regulatory_authority': [14.0, 15.0, 15.0, 16.0, 16.0, 17.0, 18.0],
    'regulatory_mandate': [16.5, 17.0, 17.5, 18.0, 18.5, 19.0, 19.5],
    'regulatory_regime': [20.0, 21.0, 22.0, 23.0, 23.5, 24.0, 25.0],
    'competition_framework': [18.0, 18.5, 19.0, 20.0, 20.5, 21.0, 21.5],
    'overall_ict_score': [68.5, 71.5, 73.5, 77.0, 78.5, 81.0, 84.0]
}
df = pd.DataFrame(raw_tracker_data)

df.columns = [col.strip().lower().replace(' ', '_') for col in df.columns]

numeric_cols = ['regulatory_authority', 'regulatory_mandate', 'regulatory_regime', 'competition_framework', 'overall_ict_score']
df[numeric_cols] = df[numeric_cols].astype(np.float64)
df['year'] = df['year'].astype(int)

assert df.isnull().sum().sum() == 0, 'Missing values detected!'
assert df.duplicated(subset=['iso3', 'year']).sum() == 0, 'Duplicate records detected!'

def classify_generation(score):
    if score < 40.0: return 'G1 (Monopoly)'
    elif score < 70.0: return 'G2 (Basic Reform)'
    elif score < 85.0: return 'G3 (Enabling Investment)'
    else: return 'G4 (Integrated Regulation)'

df['regulatory_generation'] = df['overall_ict_score'].apply(classify_generation)
df['yoy_growth_pct'] = (df['overall_ict_score'].pct_change() * 100).fillna(0.0)

df.to_csv('ict_regulatory_cleaned.csv', index=False)
print('Data Preparation Complete. Cleaned file exported.')
```
