# NDTA 631: Member 5 - NumPy and Numerical Analysis

## South Africa: ICT Regulatory Environment and Cybersecurity Readiness

This section is based on the analysis-ready information supplied by Member 4. The calculations use NumPy and are designed to satisfy the Numerical Analysis requirement: arrays, descriptive statistics, reshaping, array operations, percentage changes and interpretation.

**Data limitation:** The supplied ICT documentation confirms seven annual observations from 2016 to 2022 and the endpoints 68.5 and 84.0, but it does not provide every annual score in the ZIP. Therefore, this notebook does not invent the missing intermediate ICT values. The GCI documentation provides the 2020 and 2024 overall scores and several dimension values, which are used directly below.


```python
import numpy as np

# ICT Regulatory Tracker values explicitly documented by Member 4
ict_years = np.array([2016, 2022])
ict_scores = np.array([68.5, 84.0], dtype=float)

# GCI overall scores explicitly documented by Member 4
gci_years = np.array([2020, 2024])
gci_scores = np.array([71.02, 78.36], dtype=float)

# GCI dimension values explicitly documented by Member 4
gci_2020 = np.array([17.18, 15.90, 15.36, 18.84])
gci_2024 = np.array([18.90, 18.38, 17.56, 16.20])
dimension_names = np.array(["Legal", "Technical", "Capacity Development", "Cooperation"])

print("ICT scores:", ict_scores)
print("GCI scores:", gci_scores)
print("GCI dimensions 2020:", gci_2020)
print("GCI dimensions 2024:", gci_2024)
```

## 1. Descriptive statistics

```python
for name, arr in [("ICT endpoint values", ict_scores),
                  ("GCI overall scores", gci_scores),
                  ("GCI 2020 dimension values", gci_2020),
                  ("GCI 2024 dimension values", gci_2024)]:
    print("\n", name)
    print("Mean:", np.mean(arr))
    print("Median:", np.median(arr))
    print("Standard deviation:", np.std(arr))
    print("Minimum:", np.min(arr))
    print("Maximum:", np.max(arr))
```

## 2. Differences and percentage change

```python
ict_difference = np.diff(ict_scores)
ict_percentage_change = (ict_scores[-1] - ict_scores[0]) / ict_scores[0] * 100

gci_difference = np.diff(gci_scores)
gci_percentage_change = (gci_scores[-1] - gci_scores[0]) / gci_scores[0] * 100

print("ICT absolute increase:", ict_difference[0])
print("ICT percentage increase:", ict_percentage_change)

print("GCI absolute increase:", gci_difference[0])
print("GCI percentage increase:", gci_percentage_change)
```

## 3. Array reshaping and element-wise operations

```python
# 2 x 4 matrix: rows represent years, columns represent dimensions
gci_matrix = np.vstack((gci_2020, gci_2024))
print("GCI matrix:\n", gci_matrix)
print("Shape:", gci_matrix.shape)

dimension_change = gci_matrix[1] - gci_matrix[0]
dimension_pct_change = (dimension_change / gci_matrix[0]) * 100

for name, change, pct in zip(dimension_names, dimension_change, dimension_pct_change):
    print(f"{name}: {change:.2f} points, {pct:.2f}%")
```

## 4. Reshaping

```python
# Convert the 2D GCI matrix into one column, then reshape it back
flattened = gci_matrix.flatten()
reshaped = flattened.reshape(2, 4)

print("Flattened shape:", flattened.shape)
print("Reshaped shape:", reshaped.shape)
print(reshaped)
```

## 5. Normalisation

```python
def min_max_normalise(arr):
    return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print("Normalised ICT endpoint values:", min_max_normalise(ict_scores))
print("Normalised GCI overall values:", min_max_normalise(gci_scores))
print("Normalised 2024 GCI dimensions:", min_max_normalise(gci_2024))
```

## 6. Numerical interpretation

The ICT Regulatory Tracker increased by 15.5 points between the documented 2016 and 2022 endpoints, which is approximately a 22.63% increase. This supports the conclusion that South Africa's ICT regulatory environment improved over the study period.

The GCI overall score increased by 7.34 points between 2020 and 2024, which is approximately a 10.34% increase. The improvement was uneven across the dimensions available in the documentation. Technical capability showed the largest increase of 2.48 points, Capacity Development increased by 2.20 points, Legal increased by 1.72 points, while Cooperation declined by 2.64 points.

Because the supplied ICT documentation does not provide the exact 2020 ICT overall score in the ZIP, a direct 2020-to-2024 numerical correlation between the two datasets should not be calculated from this package alone. The findings should therefore be described as complementary evidence rather than proof of causation.
